Data Cleaning


```python
cleaned_data = {}

for sheet in excel.sheet_names:
    df = pd.read_excel(excel_path, sheet_name=sheet)

    # Hapus duplikat
    df = df.drop_duplicates()

    # Rapikan nama kolom
    df.columns = df.columns.str.strip()

    # Ubah string kosong menjadi NaN
    df = df.replace(r'^\s*$', pd.NA, regex=True)

    cleaned_data[sheet] = df

print("Data cleaning selesai.")
```

Simpan Hasil Cleaning

```python
output_path = "/content/drive/MyDrive/Hackathon AI /cleaned_data/clean_dataset.xlsx"

with pd.ExcelWriter(output_path) as writer:
    for sheet, df in cleaned_data.items():
        df.to_excel(writer, sheet_name=sheet, index=False)

print("Clean dataset berhasil disimpan.")
```

```python
import pandas as pd
import numpy as np

df = pd.read_excel(
    excel_path,
    sheet_name="Info Seputar Danau Toba (TOP 3)",
    header=None
)
```

```python
# Ambil 3 baris header
header = df.iloc[:3].copy()

# Isi merged cell yang kosong ke kanan
header = header.ffill(axis=1)

# Isi merged cell yang kosong ke bawah
header = header.ffill(axis=0)

header
```

```python
new_columns = []

for col in range(header.shape[1]):
    parts = []

    for row in range(3):
        value = str(header.iloc[row, col]).strip()

        if value != "" and value.lower() != "nan":
            parts.append(value)

    # Hilangkan duplikasi
    parts = list(dict.fromkeys(parts))

    new_columns.append(" | ".join(parts))

new_columns
```

```python
info_df = df.iloc[3:].reset_index(drop=True)

info_df.columns = new_columns
```

```python
info_df.columns = (
    info_df.columns
        .str.replace("\n"," ", regex=False)
        .str.replace("  "," ", regex=False)
        .str.strip()
)
```

```python
info_df["Nama Kabupaten"] = info_df["Nama Kabupaten"].ffill()
```

```python
info_df = info_df.dropna(how="all")

info_df = info_df.reset_index(drop=True)
```

```python
cleaned_data["Info Seputar Danau Toba (TOP 3)"] = info_df
```
