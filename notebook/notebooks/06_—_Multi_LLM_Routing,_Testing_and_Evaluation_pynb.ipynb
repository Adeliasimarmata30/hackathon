FORMAT CONTEXT

```python
import re

def format_context(contexts):

    formatted = []

    for doc in contexts:

        content = doc["content"]

        place = re.search(r"place-name:\s*(.*)", content)
        rating = re.search(r"reviewer-rating:\s*(.*)", content)
        review = re.search(r"review-text:\s*(.*)", content)

        place = place.group(1).strip() if place else "Tidak diketahui"
        rating = rating.group(1).strip() if rating else "-"
        review = review.group(1).strip() if review else "-"

        formatted.append(f"""
Nama Tempat : {place}

Rating : {rating}

Ulasan :
{review}
""")

    return "\n\n---------------------------\n\n".join(formatted)
```

Fungsi Retrieval

```python
def retrieve_context(question, top_k=5):

    query_embedding = embedding_model.encode(
        question,
        normalize_embeddings=True
    ).tolist()

    response = supabase.rpc(
        "match_documents",
        {
            "query_embedding": query_embedding,
            "match_count": top_k,
            "match_threshold": 0.55
        }
    ).execute()

    if response.data is None:
        return []

    return response.data
```

Build Prompt

```python
def build_prompt(question, contexts):

    context = format_context(contexts)

    prompt = f"""
{SYSTEM_PROMPT}

========================================================
CONTEXT
========================================================

{context}

========================================================
QUESTION
========================================================

{question}

========================================================
INSTRUCTIONS
========================================================

Before answering:

1. Read all available information carefully.

2. Combine similar information.

3. Never repeat the same information.

4. Never copy the review exactly.

5. Rewrite naturally.

6. Explain like a professional tourism consultant.

7. If multiple places are relevant:

- Compare them briefly.

- Explain why they are recommended.

8. Mention ratings naturally when available.

9. Never mention the words:

Context
Database
RAG
Embedding
Metadata

10. If multiple recommendations exist, ALWAYS use "-" bullet points.

11. Finish with a helpful recommendation.

========================================================
ANSWER
========================================================
"""

    return prompt
```

Chat Function

```python
def ask_toba_ai(question):

    # =====================================================
    # STEP 1 : RETRIEVE CONTEXT
    # =====================================================
    contexts = retrieve_context(question)

    # =====================================================
    # STEP 2 : CHECK RAG
    # =====================================================
    if len(contexts) == 0:

        return (
            "Maaf, saya tidak menemukan informasi yang relevan mengenai pertanyaan tersebut pada basis pengetahuan Toba Infinity.\n\n"
            "Saya dirancang khusus untuk membantu informasi mengenai:\n"
            "- Destinasi wisata Danau Toba\n"
            "- Hotel dan penginapan\n"
            "- Kuliner khas\n"
            "- Budaya Batak\n"
            "- Transportasi\n"
            "- Itinerary perjalanan\n\n"
            "Silakan ajukan pertanyaan lain yang berkaitan dengan kawasan Danau Toba."
        )

    # =====================================================
    # STEP 3 : BUILD PROMPT
    # =====================================================
    prompt = build_prompt(question, contexts)

    # =====================================================
    # STEP 4 : GEMINI (PRIMARY)
    # =====================================================
    try:

        response = gemini.models.generate_content(
            model=PRIMARY_MODEL,
            contents=prompt
        )

        return response.text

    except Exception as e:

        print("\n========== GEMINI ERROR ==========")
        print(e)

    # =====================================================
    # STEP 5 : GROQ (FALLBACK 1)
    # =====================================================
    try:

        response = groq.chat.completions.create(
            model=GROQ_MODEL,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0.2
        )

        return response.choices[0].message.content

    except Exception as e:

        print("\n========== GROQ ERROR ==========")
        print(e)

    # =====================================================
    # STEP 6 : OPENROUTER (FALLBACK 2)
    # =====================================================
    try:

        response = openrouter.chat.completions.create(
            model=OPENROUTER_MODEL,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0.2
        )

        return response.choices[0].message.content

    except Exception as e:

        print("\n========== OPENROUTER ERROR ==========")
        print(e)

    # =====================================================
    # STEP 7 : ALL FAILED
    # =====================================================
    return (
        "Maaf, layanan AI sedang mengalami gangguan sementara.\n\n"
        "Silakan coba kembali beberapa saat lagi."
    )
```

RAG EVALUATION

```python
# ==========================================================
# RAG EVALUATION
# ==========================================================

evaluation_questions = [

    "Hotel terbaik di Balige",

    "Tempat wisata yang wajib dikunjungi di Balige",

    "Kuliner khas Balige",

    "Apa itu Tari Tor-Tor?",

    "Apa itu Ulos?",

    "Bagaimana menuju Pulau Samosir?",

    "Buat itinerary 2 hari di Danau Toba",

    "Tempat melihat sunset di Danau Toba",

    "Siapa Presiden Amerika Serikat tahun 1990?"
]
```

```python
import pandas as pd
import time

results = []

for question in evaluation_questions:

    print("="*80)
    print(question)
    print("="*80)

    start = time.time()

    contexts = retrieve_context(question)

    answer = ask_toba_ai(question)

    end = time.time()

    response_time = round(end-start,2)

    results.append({

        "Question": question,

        "Context Found": len(contexts),

        "Response Time (s)": response_time,

        "Answer": answer
    })

evaluation_df = pd.DataFrame(results)

evaluation_df
```

```python
print("="*60)

print("RAG EVALUATION SUMMARY")

print("="*60)

print(f"Total Questions : {len(evaluation_df)}")

print(f"Average Response Time : {evaluation_df['Response Time (s)'].mean():.2f} seconds")

print(f"Average Retrieved Context : {evaluation_df['Context Found'].mean():.2f}")
```

EVALUATION SUMMARY

```python
# ==========================================================
# EVALUATION SUMMARY
# ==========================================================

total_questions = len(evaluation_df)

relevant = (evaluation_df["Relevant"] == "Yes").sum()

correct = (evaluation_df["Correct"] == "Yes").sum()

partial = (evaluation_df["Correct"] == "Partial").sum()

incorrect = (evaluation_df["Correct"] == "No").sum()

hallucination = (evaluation_df["Hallucination"] == "Yes").sum()

accuracy = ((correct + (0.5 * partial)) / total_questions) * 100

avg_response = evaluation_df["Response Time (s)"].mean()

print("="*60)
print("TOBA INFINITY RAG EVALUATION SUMMARY")
print("="*60)

print(f"Total Questions        : {total_questions}")
print(f"Relevant Answers       : {relevant}")
print(f"Correct Answers        : {correct}")
print(f"Partial Answers        : {partial}")
print(f"Incorrect Answers      : {incorrect}")
print(f"Hallucinations         : {hallucination}")
print(f"Average Response Time  : {avg_response:.2f} seconds")
print(f"Accuracy               : {accuracy:.2f}%")
```
